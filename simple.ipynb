# Simple H3 + Turbo + SelfLift

Start with `uv sync --extra notebook`, then `uv run --extra notebook jupyter lab simple.ipynb` from this folder. Select the project kernel.

No ComfyUI. This is an experimental standalone implementation, with tiny-model tests but no full pretrained render. It needs around 75 GB host RAM (128 GiB recommended), a native BF16 GPU and at least 140 GiB free model storage. The current T4/13 GB runtime is insufficient.


```python
from h3_functions import VideoSettings, inspect_runtime, load_models, apply_lora, configure_memory, generate_video
inspect_runtime("models")
```

```python
settings = VideoSettings(
    prompt="A woman gently tilts her head and smiles. Natural movement, steady camera, quiet room ambience.",
    image_path=None,  # Or "photo.jpg".
    width=544, height=960, seconds=5, seed=9174,
)
```

```python
pipe = load_models(cache_dir="models")
apply_lora(pipe, strength=settings.turbo_strength)
configure_memory(pipe)
```

```python
video = generate_video(pipe, settings, output_dir="output", use_selflift=True)
from IPython.display import Video, display
display(Video(str(video), embed=True))
```
